#### Defining Functions

```python
def greeting(name):
    print("Welcome, " + name)
    
greeting("Kay")
greeting("Cameron")
```

```python
def greeting(name, department):
    print("Welcome, " + name)
    print("You are part of " + department)
    
greeting("Blake", "Software engineering")
greeting("Ellis", "Software engineering")
```

```python
# Area function
def calculate(d):
    q = 3.14
    z = q * (d ** 2)
    print(z)

calculate(5)
#Output is 78.5
```

```python
# Area function
def circle_area(radius):
    pi = 3.14
    area = pi * (radius ** 2)
    print(area)

circle_area(5)
#Output is 78.5
```
